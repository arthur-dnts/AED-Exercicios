### Questão 1

Explique, com suas próprias palavras, a diferença entre uma hipótese de pesquisa e uma hipótese estatística. Apresente um exemplo de cada uma relacionado à área de computação.

**Hipótese de pesquisa** é uma afirmação substantiva, em linguagem natural, sobre um fenômeno ou relação que o pesquisador deseja investigar. Ela expressa uma crença ou expectativa sobre o mundo real, sem precisar ser matematicamente precisa.

**Hipótese estatística** é uma afirmação matematicamente precisa sobre um parâmetro populacional (como uma média ou probabilidade). Ela traduz a hipótese de pesquisa em termos quantificáveis, que podem ser testados com dados.

**Exemplo na área de computação:**

- *Hipótese de pesquisa:* "Algoritmos de ordenação baseados em comparação são mais lentos do que algoritmos não-comparativos para grandes conjuntos de dados."

- *Hipótese estatística:* Seja $\mu_1$ o tempo médio de execução do QuickSort e $\mu_2$ o tempo médio do Counting Sort para $n = 10^6$ elementos. A hipótese estatística seria: $H_0: \mu_1 = \mu_2$ versus $H_1: \mu_1 > \mu_2$.

### Questão 2

Considere a seguinte afirmação:

*"O uso de um novo método de ensino melhora o desempenho dos alunos em programação."*

a) Identifique a hipótese de pesquisa.

b) Escreva uma hipótese estatística nula ($H_0$).

c) Escreva uma hipótese estatística alternativa ($H_1$).

**a) Hipótese de pesquisa:**

O novo método de ensino de programação produz melhores resultados de aprendizagem nos alunos do que o método tradicional.

**b) Hipótese nula ($H_0$):**

Seja $\mu_{novo}$ a média das notas dos alunos ensinados com o novo método e $\mu_{trad}$ a média das notas dos alunos com o método tradicional:

$$H_0: \mu_{novo} \leq \mu_{trad}$$

Ou seja, o novo método não produz melhora significativa no desempenho.

**c) Hipótese alternativa ($H_1$):**

$$H_1: \mu_{novo} > \mu_{trad}$$

O novo método produz, de fato, desempenho superior ao método tradicional.

### Questão 3

Em um experimento de previsão de resultados de lançamentos de moeda, um pesquisador acredita que os participantes possuem uma habilidade especial para prever corretamente o resultado.

Considere que $\theta$ representa a probabilidade de acerto.

a) Escreva a hipótese nula.

b) Escreva a hipótese alternativa para um teste bilateral.

c) Escreva a hipótese alternativa para um teste unilateral à direita.

**a) Hipótese nula:**

$$H_0: \theta = 0{,}5$$

Sem habilidade especial, os participantes acertam por mero acaso, ou seja, com probabilidade igual à de uma moeda justa.

**b) Hipótese alternativa — teste bilateral:**

$$H_1: \theta \neq 0{,}5$$

O desempenho dos participantes é diferente do acaso, seja para mais ou para menos. Usada quando não se sabe a priori a direção do efeito.

**c) Hipótese alternativa — teste unilateral à direita:**

$$H_1: \theta > 0{,}5$$

Os participantes acertam *mais* do que o esperado pelo acaso, sugerindo uma habilidade preditiva real. Este é o teste mais adequado ao contexto, pois o pesquisador acredita especificamente que o desempenho será *acima* do acaso.

### Questão 4

Explique o que é um erro do Tipo I e o que é um erro do Tipo II.

Em seguida, descreva uma situação prática em que cada um desses erros poderia ocorrer na avaliação de um novo sistema computacional.

**Erro do Tipo I (Falso Positivo):**

Ocorre quando rejeitamos $H_0$ sendo ela verdadeira. Em outras palavras, concluímos que existe um efeito ou diferença quando, na realidade, não existe. A probabilidade de cometer esse erro é o nível de significância $\alpha$.

**Erro do Tipo II (Falso Negativo):**

Ocorre quando não rejeitamos $H_0$ sendo ela falsa. Ou seja, deixamos de detectar um efeito real que existe na população. A probabilidade de cometer esse erro é denotada por $\beta$.

---

**Situações práticas na avaliação de um novo sistema computacional:**

- **Erro Tipo I:** Um pesquisador avalia um novo sistema de detecção de intrusão e conclui, com base nos testes, que ele detecta ataques com maior precisão do que o sistema atual — quando, na verdade, a diferença observada foi apenas aleatória. A empresa investe na troca desnecessária do sistema.

- **Erro Tipo II:** O pesquisador testa o mesmo sistema, mas falha em detectar que ele é genuinamente melhor (talvez por ter usado uma amostra de testes pequena). O novo sistema é descartado, e a empresa perde a oportunidade de uma melhoria real na segurança.

### Questão 5

Considere a tabela abaixo:

| Decisão | Realidade |
|---|---|
| Rejeitar $H_0$ | $H_0$ é verdadeira |
| Não rejeitar $H_0$ | $H_0$ é falsa |

Para cada caso apresentado, indique qual tipo de erro foi cometido.

**Caso 1 — Rejeitar $H_0$ quando $H_0$ é verdadeira:**

Este é um **Erro do Tipo I**. Rejeitamos a hipótese nula, mas ela era verdadeira. É equivalente a condenar um inocente num tribunal. A probabilidade desse erro é controlada pelo nível de significância $\alpha$.

**Caso 2 — Não rejeitar $H_0$ quando $H_0$ é falsa:**

Este é um **Erro do Tipo II**. Retemos a hipótese nula, mas ela era falsa — deixamos de detectar um efeito real. É equivalente a absolver um culpado. A probabilidade desse erro é $\beta$, e o poder do teste ($1 - \beta$) mede a capacidade de evitá-lo.

### Questão 6

Explique o significado do nível de significância ($\alpha$).

Por que a literatura estatística normalmente utiliza $\alpha = 0.05$?

**Significado do nível de significância:**

O nível de significância $\alpha$ é a probabilidade máxima aceitável de cometer um **Erro do Tipo I**, ou seja, a probabilidade de rejeitar $H_0$ quando ela é verdadeira. Ele define o limiar de evidência necessário para rejeitar a hipótese nula: se a probabilidade de obter os dados observados (ou dados mais extremos), assumindo $H_0$ verdadeira, for menor que $\alpha$, rejeitamos $H_0$.

**Por que $\alpha = 0{,}05$?**

O valor $\alpha = 0{,}05$ foi popularizado por Ronald Fisher no início do século XX como um limiar prático de razoabilidade. Ele representa um equilíbrio histórico e convencional entre:

- Ser *exigente o suficiente* para evitar falsas descobertas frequentes (Tipo I);
- Ser *permissivo o suficiente* para que estudos com amostras moderadas ainda possam detectar efeitos reais.

Não há uma justificativa matemática absoluta para esse valor — trata-se de uma convenção amplamente adotada. Em contextos de maior risco (medicina, segurança de sistemas), valores mais rígidos como $\alpha = 0{,}01$ ou $\alpha = 0{,}001$ são preferidos.

### Questão 7

O poder de um teste é definido como $1 - \beta$.

a) O que representa $\beta$?

b) O que representa o poder do teste?

c) Por que um pesquisador deseja testes com alto poder estatístico?

**a) O que representa $\beta$:**

$\beta$ é a probabilidade de cometer um **Erro do Tipo II**: não rejeitar $H_0$ quando ela é de fato falsa. É a chance de o teste "deixar escapar" um efeito real que existe na população.

**b) O que representa o poder do teste:**

O poder do teste, $1 - \beta$, é a probabilidade de **rejeitar corretamente** $H_0$ quando ela é falsa — ou seja, a capacidade do teste de detectar um efeito real quando ele existe. Um poder de $0{,}80$, por exemplo, significa que o teste tem 80% de chance de detectar o efeito caso ele realmente exista.

**c) Por que um pesquisador deseja testes com alto poder:**

Um pesquisador deseja alto poder porque isso reduz a chance de uma conclusão negativa falsa. Se um novo algoritmo de compressão realmente for mais eficiente, um teste com alto poder garantirá que essa melhoria seja detectada. Testes com baixo poder desperdiçam recursos e podem descartar inovações genuínas. Em geral, a literatura recomenda $1 - \beta \geq 0{,}80$ como valor mínimo aceitável.

### Questão 8

Observe o código Python abaixo:

```python
from numpy import random
dados = random.binomial(
    n=100,
    p=0.5,
    size=10000
)
print(dados[:10])
```

a) Qual distribuição está sendo simulada?

b) O que representa o parâmetro `n`?

c) O que representa o parâmetro `p`?

d) Qual o objetivo dessa simulação no contexto de testes de hipóteses?

**a) Distribuição simulada:**

Está sendo simulada a **distribuição binomial** — a distribuição de probabilidade do número de sucessos em $n$ tentativas independentes, cada uma com probabilidade $p$ de sucesso.

**b) Parâmetro `n`:**

`n=100` representa o **número de tentativas** (ou ensaios) em cada experimento. No contexto do experimento ESP do livro, seria o número de participantes testados em cada rodada.

**c) Parâmetro `p`:**

`p=0.5` representa a **probabilidade de sucesso** em cada tentativa individual. Aqui, reflete a hipótese nula: sem habilidade especial, cada participante tem 50% de chance de acertar (equivalente ao acaso numa moeda justa).

**d) Objetivo da simulação:**

A simulação gera a **distribuição amostral** do teste sob $H_0$. Ao repetir o experimento 10.000 vezes, obtemos empiricamente a distribuição de quantos acertos esperaríamos se a hipótese nula fosse verdadeira. Isso nos permite identificar a **região crítica**: os valores de $X$ tão extremos que seria improvável observá-los se $H_0$ fosse verdadeira, levando à sua rejeição.

````python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

dados = random.binomial(
    n=100,
    p=0.5,
    size=10000
)
print("Primeiros 10 valores:", dados[:10])
print(f"Média: {dados.mean():.2f}")
print(f"Desvio padrão: {dados.std():.2f}")

# Visualização da distribuição amostral sob H0
sns.histplot(dados, bins=20, binwidth=0.5, color='steelblue')
plt.title("Distribuição amostral sob H₀ (θ = 0.5, n=100)")
plt.xlabel("Número de acertos (X)")
plt.ylabel("Frequência")
plt.axvline(dados.mean(), color='red', linestyle='--', label=f'Média = {dados.mean():.1f}')
plt.legend()
plt.tight_layout()
plt.show()
````

### Questão 9

Escreva um código Python que gere 5.000 observações de uma distribuição binomial com $n = 50$ e $p = 0.5$. Em seguida:

a) Calcule a média dos resultados simulados.

b) Compare a média observada com o valor esperado teoricamente.

````python
import numpy as np

# Gerar 5000 observações de uma distribuição binomial com n=50 e p=0.5
np.random.seed(42)
observacoes = np.random.binomial(n=50, p=0.5, size=5000)

# a) Calcular a média dos resultados simulados
media_simulada = np.mean(observacoes)
print(f"a) Média simulada: {media_simulada:.4f}")

# b) Comparar com o valor esperado teórico
# Para uma distribuição binomial: E[X] = n * p
valor_esperado = 50 * 0.5
print(f"b) Valor esperado teórico (n × p = 50 × 0.5): {valor_esperado:.4f}")
print(f"   Diferença (erro): {abs(media_simulada - valor_esperado):.4f}")
print()
print("Conclusão: A média simulada converge para o valor teórico esperado,")
print("confirmando que a simulação reflete corretamente a distribuição binomial.")
````

### Questão 10

Explique o conceito de estatística de teste.

Por que a estatística de teste é importante para a tomada de decisão em um teste de hipóteses?

**Conceito de estatística de teste:**

A estatística de teste é um valor numérico calculado a partir dos dados da amostra que resume a evidência contra a hipótese nula. Ela transforma os dados brutos em um único número que pode ser comparado com a distribuição amostral esperada sob $H_0$.

Por exemplo, no experimento ESP: se $N = 100$ pessoas participam e $X = 62$ acertam, a estatística de teste é o próprio valor $X = 62$ (ou, em testes mais sofisticados, um valor padronizado como o $z$-score).

**Importância para a tomada de decisão:**

A estatística de teste é o mecanismo central do teste de hipóteses porque:

1. **Define a evidência:** quantifica o quanto os dados observados se afastam do que $H_0$ prevê;
2. **Permite comparação:** ao conhecer sua distribuição amostral sob $H_0$, podemos calcular a probabilidade de observar um valor tão extremo por acaso (o valor-p);
3. **Delimita a região crítica:** valores da estatística de teste além do valor crítico levam à rejeição de $H_0$;
4. **Padroniza a decisão:** remove a subjetividade, tornando o critério de rejeição explícito e reprodutível.

### Questão 11

Explique a diferença entre teste unilateral e teste bilateral.

Apresente um exemplo de situação em que cada um deles seria mais apropriado.

**Teste bilateral (bicaudal):**

A região crítica está dividida em **dois lados** da distribuição. Rejeita-se $H_0$ tanto por valores muito altos quanto muito baixos da estatística de teste. A hipótese alternativa é da forma $H_1: \theta \neq \theta_0$.

*Exemplo:* Um engenheiro deseja saber se uma atualização de firmware **alterou** (para melhor ou pior) o tempo médio de resposta de um servidor. Como qualquer desvio é relevante, usa-se um teste bilateral.

---

**Teste unilateral (unicaudal):**

A região crítica está concentrada em **apenas um lado** da distribuição. Usa-se quando o pesquisador tem uma expectativa direcional bem fundamentada. A hipótese alternativa é $H_1: \theta > \theta_0$ (direita) ou $H_1: \theta < \theta_0$ (esquerda).

*Exemplo:* Uma equipe de desenvolvimento afirma que seu novo compilador **reduz** o tempo de compilação. Como só interessa detectar uma redução (e não um aumento), o teste unilateral à esquerda é o mais adequado e oferece maior poder para detectar esse efeito específico.

### Questão 13

Analise as hipóteses abaixo:

$$H_0: \theta = 0{,}5$$
$$H_1: \theta > 0{,}5$$

a) O teste é unilateral ou bilateral?

b) Em qual lado da distribuição estará a região crítica?

c) Como essa escolha afeta a interpretação dos resultados?

**a) O teste é unilateral (unicaudal à direita).**

A hipótese alternativa especifica uma direção clara: $\theta > 0{,}5$. Apenas valores superiores a 0,5 constituem evidência contra $H_0$.

**b) A região crítica estará no lado direito da distribuição.**

Valores grandes da estatística de teste (muitos acertos, bem acima de 50%) caem na região crítica. Por exemplo, com $\alpha = 0{,}05$ e $N = 100$, rejeitaríamos $H_0$ se $X \geq 59$ (aproximadamente), concentrando todo o 5% no lado direito.

**c) Como essa escolha afeta a interpretação:**

- **Maior poder para detectar o efeito esperado:** Ao concentrar toda a área $\alpha$ em um só lado, o teste se torna mais sensível para detectar valores de $\theta$ acima de 0,5 do que um teste bilateral com o mesmo $\alpha$.
- **Impossibilidade de detectar o efeito oposto:** Se os dados mostrassem $\theta < 0{,}5$ (participantes acertando menos do que o acaso), o teste unilateral à direita não rejeitaria $H_0$, mesmo que o resultado fosse estatisticamente notável.
- **Compromisso com a direcionalidade:** A escolha do teste unilateral deve ser feita *antes* da coleta dos dados, com base em justificativa teórica, para evitar viés de confirmação.
