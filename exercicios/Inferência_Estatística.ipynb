### Questão 3 — Erro Amostral

Defina o conceito de erro amostral.

- As duas amostras podem apresentar médias distintas por se tratarem de subconjuntos distintos de um conjunto maior, cada amostra possui seus próprios elementos, podendo ser uma amostra mais homogênea, enquanto a outra mais variável, a presença de outliers também podem influenciar no resultado final da média.

- Erro amostral é a diferença entre a estatística amostral e o parâmetro populacional, é um desvio natural que surge após a estimativa em parte dos dados e não no conjunto completo.

### Questão 4 — Simulação de Médias Amostrais

Utilize Python para:
- gerar uma população com distribuição normal;
- extrair 100 amostras aleatórias de tamanho 30;
- calcular a média de cada amostra.
Depois:

a. armazene as médias em uma lista;

b. calcule a média das médias amostrais.

````python
import numpy as np

populacao = np.random.normal(loc=50, scale=10, size=100000)

n_amostras = 100
tamanho_amostra = 30

medias_amostrais = []

for _ in range(n_amostras):
    amostra = np.random.choice(populacao, size=tamanho_amostra, replace=False)
    medias_amostrais.append(np.mean(amostra))

print("Média das médias:", np.mean(medias_amostrais))
print("Desvio padrão das médias:", np.std(medias_amostrais))
````

### Questão 7 — Erro Padrão
Considere o conjunto de dados:

```python
import numpy as np
dados = [5, 8, 9, 10, 12, 15, 18]
```

a. Calcule o desvio padrão amostral.

b. Calcule o erro padrão da média.

Utilize Python em sua resposta.

````python
dados = [5, 8, 9, 10, 12, 15, 18]

desvio = np.std(dados)
print(f"Desvio-Padrão: {desvio}")

erro_padrao = desvio/np.sqrt(len(dados))
print(f"Erro Padrão: {erro_padrao}")
````

### Questão 8 — Interpretação do Erro Padrão

Explique com suas palavras a relação entre tamanho da amostra e erro padrão.

- A relação entre o tamanho e o erro padrão é que o erro padrão é calculado a partir da razão entre o desvio-padrão e a raiz quadrada do tamanho da amostra, ou seja, o erro é relativo à quantidade de elementos da amostra, ou seja, quanto maior o tamanho da amostra, menor o erro.

### Questão 9 — Teorema Central do Limite

Escreva um código em Python que:
- gere valores aleatórios de uma distribuição exponencial;
- extraia várias amostras;
- calcule as médias das amostras;
- construa um histograma das médias.

O que você observa no formato do histograma?

````python
import matplotlib.pyplot as plt
import numpy as np

dados = np.random.exponential(scale=1/0.5, size=1000)
n_amostras = 100
tamanho = 30

medias_amostrais = []

for _ in range(n_amostras):
    amostra = np.random.choice(dados, tamanho, replace=True)
    medias_amostrais.append(np.mean(amostra))

plt.hist(medias_amostrais, bins=20)
plt.title("Distribuição das Médias Amostrais")
plt.xlabel("Média")
plt.ylabel("Frequência")
plt.show()
````

- É possível inferir que a distribuição amostral das médias é normal, os valores tendem a se concentrar em torno da verdadeira média populacional e há poucos valores extremos (outliers).

### Questão 11 — Intervalo de Confiança com SciPy

Utilize as bibliotecas numpy e scipy.stats para calcular um intervalo de confiança para a média dos
dados:

dados = [22, 25, 21, 20, 26, 24, 23, 27]

````python
import numpy as np
import scipy.stats

dados = [22, 25, 21, 20, 26, 24, 23, 27]

media = np.mean(dados)
desvio = np.std(dados)
erro_padrao = desvio/np.sqrt(len(dados))
sem = scipy.stats.sem(dados)

ic = scipy.stats.t.interval(confidence=0.95, df=len(dados)-1, loc=media, scale=sem)

print("Média:", media)
print("Intervalo de confiança 95%:", (f"{float(ic[0]):.2f}", f"{float(ic[1]):.2f}"))
````

### Questão 13 — Variabilidade das Médias
Escreva um programa em Python que:
- gere 50 amostras aleatórias de tamanho 5;
- gere 50 amostras aleatórias de tamanho 100;
- compare a variabilidade das médias obtidas.

O que acontece quando o tamanho da amostra aumenta?

````python
import matplotlib.pyplot as plt
import numpy as np

dados = np.random.exponential(scale=1/0.5, size=1000)
n_amostras = 50
tamanho_a = 5
tamanho_b = 100

medias_a = []
medias_b = []

for _ in range(n_amostras):
    amostra = np.random.choice(dados, tamanho_a, replace=True)
    medias_a.append(np.mean(amostra))

for _ in range(n_amostras):
    amostra = np.random.choice(dados, tamanho_b, replace=True)
    medias_b.append(np.mean(amostra))

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(medias_a, bins=20)
plt.title("Médias Amostrais - A = 5")
plt.xlabel("Média")
plt.ylabel("Frequência")

plt.subplot(1, 2, 2)
plt.hist(medias_b, bins=20)
plt.title("Médias Amostrais - B = 100")
plt.xlabel("Média")
plt.show()
````

- Ao aumentar o tamanho da amostra, a média tende a se concentrar mais próxima do valor real da média populacional

- estimativa intervalar.

- A estimativa pontual é a utilização de um único ponto ou valor para estimar um parâmetro desconhecido da população.
