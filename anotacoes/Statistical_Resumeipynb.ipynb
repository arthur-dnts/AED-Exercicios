### O que é estatística?

- É a prática da coleta e do estudo de dados

- Resumo estatístico: Um fato ou resumo de algum dado

- O que a estatística pode fazer?
  - Qual é a probabilidade de alguém comprar um produto?
  - É mais provável que comprem o produto se puderem usar um meio de pagamento diferente?
  - Quantos clientes sua empresa terá? Como otimizar a gestão?
  - Quantos tamanhos de jeans precisam se fabricados para que sirvam em 95% da população? Deve ser produzido o mesmo número de cada tamanho?

- O que a estatística não faz:
  - Por que Game of Thrones é tão popular?
    - Ao invés disso podemos nos perguntar: Séries com mais violência são vistas por mais pessoas? Mas mesmo assim, isso não diz se mais violência leva a mais visualizações

- Há dois tipos de estatísticas:
  1. Estatísticas descritivas
    - Descreve e resume os dados
    - Ex: 50% dos trabalhadores usam veículos particulares, 25% usam transporte coletivo e 25% vão de bicicleta ao trabbalho
  2. Estatísticas inferenciais
    - Usa uma amostra de dados para fazer inferências sobre um grupo maior
    - Ex: Porcentagem de pessoas que dirigem ao trabalho

- Tipos de dados:
  - Numéricos (Quantitativos)
    - Contínuo (medido)
      - Ex: Velocidade de um veículo, tempo de espera na fila
    - Discreto (contado)
      - Ex: Número de pets, número de pacotes enviados

  - Categóricos (Qualitativos)
    - Nominal (não ordenado)
      - Ex: Casados/Não casados, país de residência
    - Ordinal (ordenado)
      - Ex: Discordo totalmente, Discordo em partes, Não concordo nem discordo, etc

- Dados categóricos podem ser representados por números (1 e 0 para Verdadeiro ou Falso e número naturais para pesquisas de satisfação, por exemplo)

### Medidas de centro

- A média (avarage) é uma das formas mais comuns de resumir dados

- No cálculo da média, todos os valores são somados e depois divididos pela quantidade total de elementos

- Também é possível calcular a média a partir da biblioteca NumPy do Python:

```python
media = np.mean(df["coluna"])
```

- A mediana é a medida do centro dos dados

- Para calcular a mediana os dados ordenados (crescente ou decrescente) e o valor central será a mediana. Em casos em que o conjunto de dados é par, uma média entre os dois termos centrais deve ser efetuada para gerar a mediana

- Também é possível calcular a mediana a partir da biblioteca NumPy do Python:

```python
media = np.median(df["coluna"])
```

- A moda é basicamente o valor mais comum (que se repete) no conjunto de dados. A moda é frequentemente utilizada em variáveis categóricas

- Também é possível calcular a média a partir da biblioteca NumPy do Python:

```python
media = np.mode(df["coluna"])
```

- Qual medida usar:
  - Para casos de dados simétricos, a média é o mais comum
  - Para dados enviesados (não simétricos, ou seja, com valores discrepantes) é necessário utilizar a mediana


### Medidas de dispersão

- A dispersão descreve a distância ou proximidade dos pontos de dados

- A variância mede a distância média de cada ponto de dados para a média dos dados
  - A variância é calculada subtraindo a média de cada ponto de dados, em seguida cada distância é elevada ao quadrado e depois somadas, por fim, a soma das distâncias quadradas é dividida pelo número de pontos de dados - 1:
  ```python
  dists = df["coluna"] - np.mean(df["coluna"])

  sq_dists = dists ** 2

  sum_sq_dists = = np.sum(sq_dists)

  variance = sum_sq_dists / (total - 1)
  ```
  - A variância também pode ser calculada automaticamente a partir da biblioteca NumPy. O argumento ddof pode receber o valor 0 ou 1, que representam, respectivamente, a variância da população (conjunto de dados completo) e a variância da amostra
  ```python
  np.var(df["coluna"], ddof=1)
  ```

- O Desvio padrão é uma outra medida de dispersão, calculado a partir da raiz quadrada da variância. Seu uso é mais comum porque não há alteração na unidade de medida dos dados, pode ser usado a partir da biblioteca NumPy
```python
np.sqrt(np.var(df["coluna"], ddof=1))

# ou

np.std(df["coluna"], ddof=1)
```

- O desvio médio absoluto considera o valor absoluto das distâncias em relação a média e em seguida considera a média dessas diferenças

```python
dists = df["coluna"] - np.mean(df["coluna"])
np.mean(np.abs(dists))
```

  - Desvio padrão vs. Disvio médio abolsuto
    - O desvio padrão eleva as distâncias ao quadrado, penalizando distâncias longas mais do que curtas
    - O desvio médio absoluto penaliza as distâncias igualmente

- Os Quantis dividem os dados em um números de partes iguais. Podem ser calculados a partir da biblioteca NumPy:

```python
np.quantile(df["coluna"], [0, 0.25, 0.5, 0.75, 1])
# 0.5 = mediana

# Também é possível calcular usando o np.linspace
np.quantile(df["coluna"], np.linspace(0, 1, 5))
```

  - Box plots usam quartis. A linha horizontal da caixa marca o segundo quartil, a borda superior e inferior são, respectivamente, o terceiro quartil e o primeiro quartil, a altura da caixa é o intervalo interquartil (IQR)

- Discrepâncias são ponto de dados substancialmente diferente dos demais. Um ponto de dado é discrepante se: Q1-1.5*IQR < dado < Q3+1.5*IQR
```python
from scipy.stats import iqr

iqr = iqr(df["coluna"])
lower_threshold = np.quantile(df["coluna"], 0.25) - 1.5 * iqr
upper_threshold = np.quantile(df["coluna"], 0.75) + 1.5 * iqr

print(df[df["coluna"] < lower_threshold | df["coluna"] > upper_threshold])
```

```python
# Calculate total co2_emission per country: emissions_by_country
emissions_by_country = food_consumption.groupby('country')['co2_emission'].sum()

# Compute the first and third quantiles and IQR of emissions_by_country
q1 = np.quantile(emissions_by_country, 0.25)
q3 = np.quantile(emissions_by_country, 0.75)
iqr = q3 - q1

# Calculate the lower and upper cutoffs for outliers
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

# Subset emissions_by_country to find outliers
outliers = emissions_by_country[(emissions_by_country < lower) | (emissions_by_country > upper)]
print(outliers)
```
