### Introduction to summary statistics: The sample mean and median

- Para casos em que os dados devem ser resumidos de forma mais suscinta

- A média é, basicamente, a soma de todos os valores divido pela quantidade total de elementos

$$ média = \bar{x} = \frac{1}{n} \sum_{i=1}^N X_i$$

- Um ponto negativo no uso da média é que ela é facilmente manipulável por dados que se diferenciam dos demais drasticamente (outliers)

- A mediana, por outro lado, representa o valor central do conjunto de dados. Caso o conjunto seja par a mediana é definida pela média dos dois valores centrais

- Ambas podem ser calculadas a partir da biblioteca NumPy:

```python
import numpy as np

np.mean(df)
np.meadian(df)
```

### Percentiles, outliers and box plots

- O 50º percentil dos dados são menores que a mediana. O 25º percentil dos dados é maior que 25% dos dados ordenados, e assim por diante para cada percentil

- É possível calcular o percentil a partir da biblioteca NumPy:

```python
np.percentil(df["coluna"], [25, 50, 75])
```

- A lista passada como segundo argumento representa os percentis desejados, não frações

- Para plotar esses dados a melhor forma é utilizando um box plot

- O centro da caixa representa a mediana (50º percentil), a borda superior e inferior da caixa representam, respectivamente, o 75º percentil e o 25º percentil

- A altura da caixa representa 50% dos dados e é chamada de Intervalo Interquartil (IQR)

- Os bigodes são posicionados a 1.5*IQR ou até a extensão dos dados

- Os outliers são representados como bolinhas após os bigodes. É importante fixar que não é correto assumir, quando não há evidências, que os outliers são erros

### Variance and standard deviation

- A variância é a distância ao quadrado dos dados de sua média, informalmente utilizada como uma medida de dispersão de dados. Por se tratar de uma valor ao quadrado, a unidade de medida retornada não é a mesma dos dados medidos

- Para resolver o problema de antes utiliza-se o desvio padrão, que calcula a raiz quadrada da variância

- Ambas podem ser calculadas a partir da biblioteca NumPy:

```python
np.var(df["coluna"])
np.std(df["coluna"]) # ou np.sqrt(df["coluna"])
```

- Exercícios:

```python
# Array of differences to mean: differences
differences = versicolor_petal_length - np.mean(versicolor_petal_length)

# Square the differences: diff_sq
diff_sq = differences**2

# Compute the mean square difference: variance_explicit
variance_explicit = np.mean(diff_sq)

# Compute the variance using NumPy: variance_np
variance_np = np.var(versicolor_petal_length)

# Print the results
print(variance_explicit, variance_np)
```

```python
# Compute the variance: variance
variance = np.var(versicolor_petal_length)

# Print the square root of the variance
print(np.sqrt(variance))

# Print the standard deviation
print(np.std(versicolor_petal_length))
```

### Covariance and the Pearson correlation coefficient

- Plotagem de dados em um scatterplot

- Uma outra maneira de visualizar a relação entre a quantidade total de votos e os votos para Obama seria utilizando a covariância

- A covariância mede como duas quantidades variam juntas. Ou seja, ela é a média do produto das diferenças das quantidades

$$ covariância = \frac{1}{n} \sum_{i=1}^N (x_i - \bar{x})(y_i - \bar{y})  $$

- Quando ambas as variáveis analisadas estão acima ou abaixo de suas respectias médias, então há uma covariância positiva

- Quando as variáveis são inversas em relação as suas respectivas médias, então há uma covariância negativa

- Caso seja necessário uma medida mais generalizada de duas variáveis que dependem uma da outra, é necessário que os dados sejam adimensionais:

$$ p = correlação de Pearson = \frac{covariância}{(std_x)(std_y)} $$

- A correlação de Pearson pode varia de -1 (correlação negativa) a 1 (correlação positiva)

- A covariância pode ser calculada a partir da biblioteca NumPy:

```python
np.cov(x, y)
```

```python
def pearson_r(x, y):
    """Compute Pearson correlation coefficient between two arrays."""
    # Compute correlation matrix: corr_mat
    corr_mat = np.corrcoef(x, y)

    # Return entry [0,1]
    return corr_mat[0,1]

# Compute Pearson correlation coefficient for I. versicolor: r
r = pearson_r(versicolor_petal_length, versicolor_petal_width)

# Print the result
print(r)
```
